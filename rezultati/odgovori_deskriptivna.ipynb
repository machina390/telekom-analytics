## Niz računa: 900, 1.200, 1.300, 1.300, 2.100, 36.500.
Izračunaj sredinu, medijanu i mod. Koju prijavljuješ i zašto?


```python
import pandas as pd

bills=[900, 1200, 1300, 1300, 2100, 36500]
df = pd.DataFrame(data=bills,columns=['Bills'])
display(df)

print("Prosek: {} Medijana: {} Mod: {}".format(df["Bills"].mean(),df["Bills"].median(),df["Bills"].mode()))

```

Prijavio bih MOD zato sto:
- outlier nam u startu ubija smisao za aritmetičku
- čini se da je raspodela asimetrična ka desno (vrh raspodele je na desnoj strani grafika)

## Mesečni rast korisnika: +10%, +20%, −5%.
Prosečan rast aritmetički je 8,3%. Izračunaj geometrijski. Koji je tačan?

```python
user_growth = pd.DataFrame(data=[0.1,0.2,0.95],columns=["Growth"])

user_growth=user_growth.to_numpy()

import numpy as np

def geo_mean(arg):
    return np.exp(np.mean(np.log(arg)))

geometric=geo_mean(user_growth)

print(geometric*100)
```

Rezultat dobijen prema uputstvu: `https://stackoverflow.com/a/43099751`

Geometrijski rast je tačnija vrednost u odnosu na aritmetički.
